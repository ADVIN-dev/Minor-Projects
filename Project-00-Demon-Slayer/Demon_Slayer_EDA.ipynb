```python
# Treating the missing values.
# Drop the rows where columns are having minor missing values.
df.dropna(subset=['Age','Weight','Enemies','Voice_Actor_Japanese','Voice_Actor_English','Role'],inplace=True)
```

```python
# Missing value Treated.
print(df.isnull().sum())
```

```python
# separate Numerical & categorical Columns
num_cols = df.select_dtypes(include = ['int64','float64']).columns
num_cols
```

```python
cat_cols = df.select_dtypes(include = ['object','category']).columns
cat_cols
```

```python
df.hist(figsize = (15,12))
plt.show()
```

```python
# Box Plot
for col in num_cols:
  plt.figure(figsize = (6,3))
  sns.boxplot(df[col])
  plt.title(col)
  plt.show()
```

```python
exclude_cols = ['Character_ID']
num_cols = [col for col in num_cols if col not in exclude_cols] # Deleting extra columns..
print(num_cols)
```

```python
# Outlier Treatment

for col in num_cols:
  Q1 = df[col].quantile(0.25)
  Q3 = df[col].quantile(0.75)

  IQR = Q3-Q1

  lower_range = Q1 - 1.5*IQR
  upper_range = Q3 + 1.5*IQR

  df[col] = np.where(df[col] > upper_range, upper_range, df[col])
  df[col] = np.where(df[col]< lower_range, lower_range, df[col])
```

```python
# verifying

for col in num_cols:
  plt.figure(figsize =(6,3))
  sns.boxplot(df[col])
  plt.title(col)
  plt.show()
```

```python
print(df['Age'].value_counts())
```

```python
# Correlation Analysis
corr = df.corr(numeric_only=True)
plt.figure(figsize =(10,8))
sns.heatmap(corr, annot = True, cmap = 'coolwarm') # annot gives numerical representation inside the graph.
plt.show()                                         # cmap stands for Colormap — it controls the color scheme of the heatmap.
```

```python
# scatter plot
#to understand the relationship b/w 2 numerical values

plt.figure(figsize=(8,5))
sns.scatterplot(x = 'Age', y = 'Character_ID', hue='Race', data = df, s= 100, alpha = 0.8)
plt.title('Character_ID vs Age')
plt.show()
```

```python
# categorical analysis

plt.figure(figsize=(8,5))
sns.countplot(x = 'Status', data = df, palette = 'Set2')
plt.title('Dead or Alive')
plt.xticks(rotation = 90)
plt.show()
```
